# QBF Transformer — Wissenschaftliche Analyse
## Einfluss von Transformationen auf QBF-Solver-Verhalten

**Ziel:** Analyse wie wahrheitswerterhaltende Transformationen (Operatoren) die  
Laufzeit und Code-Coverage von DepQBF und CAQE verändern.

**Struktur:**
1. Setup & Daten laden
2. Datenqualität prüfen
3. Primäranalyse — Laufzeit-Ratio
4. Primäranalyse — Coverage-Delta
5. Sekundäranalyse — Mechanismen & Solver-Vergleich
6. Sekundäranalyse — Zusammenhang Coverage vs. Laufzeit
7. Aggregierte Ergebnistabelle (Thesis-fertig)

## 1. Setup & Daten laden

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import seaborn as sns
from scipy import stats
from pathlib import Path
import warnings
warnings.filterwarnings("ignore")

# ── Plot-Einstellungen ────────────────────────────────────────────────────
plt.rcParams.update({
    "figure.dpi":       120,
    "savefig.dpi":      300,
    "font.size":        12,
    "axes.titlesize":   13,
    "axes.labelsize":   12,
    "axes.spines.top":  False,
    "axes.spines.right":False,
    "figure.facecolor": "white",
})
sns.set_palette("muted")

# ── Operator-Konfiguration ────────────────────────────────────────────────
COLORS = {
    "tautology_insertion":              "#4C72B0",
    "pure_literal_gen_auto":            "#DD8452",
    "pure_literal_gen_pos":             "#E8A070",
    "pure_literal_gen_neg":             "#C4633A",
    "universal_reduction_trigger":      "#55A868",
    "blocked_clause_insertion":         "#C44E52",
    "blocked_clause_insertion_neg":     "#D4706E",
    "blocked_clause_insertion_v1":      "#A83538",
}

LABELS = {
    "tautology_insertion":              "Tautologie",
    "pure_literal_gen_auto":            "Pure Lit. (auto)",
    "pure_literal_gen_pos":             "Pure Lit. (pos)",
    "pure_literal_gen_neg":             "Pure Lit. (neg)",
    "universal_reduction_trigger":      "Universal Red.",
    "blocked_clause_insertion":         "Blocked Clause",
    "blocked_clause_insertion_neg":     "Blocked Clause (neg)",
    "blocked_clause_insertion_v1":      "Blocked Clause (v2)",
}

MECHANISMS = {
    "tautology_insertion":              "Tautologie-Preprocessing",
    "pure_literal_gen_auto":            "Pure Literal Elimination",
    "pure_literal_gen_pos":             "Pure Literal Elimination",
    "pure_literal_gen_neg":             "Pure Literal Elimination",
    "universal_reduction_trigger":      "Universal Reduction",
    "blocked_clause_insertion":         "Blocked Clause Addition (BCE)",
    "blocked_clause_insertion_neg":     "Blocked Clause Addition (BCE)",
    "blocked_clause_insertion_v1":      "Blocked Clause Addition (BCE)",
}

OP_ORDER = list(COLORS.keys())
print("✓ Setup abgeschlossen")
```

```python
# ── CSV laden ────────────────────────────────────────────────────────────
CSV_PATH = "results/experiments.csv"   # <-- Pfad anpassen wenn noetig

df_raw = pd.read_csv(CSV_PATH)
print(f"Geladen: {len(df_raw)} Zeilen, {df_raw['operator'].nunique()} Operatoren")
print(f"Spalten: {list(df_raw.columns)}")
df_raw.head(3)
```

## 2. Datenqualität prüfen

```python
# ── Überblick ────────────────────────────────────────────────────────────
print("=" * 60)
print("DATENQUALITÄT")
print("=" * 60)

total = len(df_raw)
equiv = df_raw["equivalent"].sum()
n_error_orig = (df_raw.get("depqbf_status_original", pd.Series()) == "ERROR").sum()

print(f"Gesamt Experimente:        {total}")
print(f"Aequivalent (korrekt):     {equiv}  ({equiv/total*100:.1f}%)")
print(f"NICHT-aequivalent:         {total-equiv}  ({(total-equiv)/total*100:.1f}%)")
print(f"  davon Original=ERROR:    {n_error_orig}  (Solver-Fehler, nicht Operator-Bug)")
print()

# Pro Operator
summary = df_raw.groupby("operator").agg(
    n=("equivalent","count"),
    n_equiv=("equivalent","sum"),
    n_skip=("equivalent", lambda x: (x==False).sum()),
).reset_index()
summary["aequiv_rate"] = (summary["n_equiv"] / summary["n"] * 100).round(1)
summary["operator_label"] = summary["operator"].map(LABELS).fillna(summary["operator"])
print(summary[["operator_label","n","n_equiv","n_skip","aequiv_rate"]]
      .to_string(index=False))
```

```python
# ── Nur gueltige (aequivalente) Experimente fuer Analyse ─────────────────
# Original=ERROR Faelle: Solver konnte Formel nicht loesen — kein Operator-Bug
# Diese werden herausgefiltert da kein valider Vergleich moeglich

df = df_raw[
    (df_raw["equivalent"] == True) &
    (df_raw["valid"] == True)
].copy()

# Operator-Reihenfolge fixieren
df["operator"] = pd.Categorical(df["operator"], categories=OP_ORDER, ordered=True)
df = df.sort_values("operator")

print(f"Analyse-Datensatz: {len(df)} Experimente (nach Filterung)")
print(f"Operatoren:")
for op, n in df.groupby("operator", observed=True).size().items():
    label = LABELS.get(op, op)
    print(f"  {label:<30} n={n}")
```

## 3. Primäranalyse — Laufzeit-Ratio

**Kennzahl:** `runtime_ratio = t_Mutant / t_Original`  
- `< 1.0` → Mutant schneller (Operator hilft dem Solver)  
- `> 1.0` → Mutant langsamer  
- `= 1.0` → kein Unterschied

```python
# ── Deskriptive Statistik ─────────────────────────────────────────────────
stats_depqbf = df.groupby("operator", observed=True)["depqbf_time_ratio"].agg([
    "count", "median", "mean", "std",
    ("q25", lambda x: x.quantile(0.25)),
    ("q75", lambda x: x.quantile(0.75)),
    ("min", "min"), ("max", "max")
]).round(4)

stats_depqbf.index = stats_depqbf.index.map(LABELS)
print("Laufzeit-Ratio DepQBF — Deskriptive Statistik:")
print(stats_depqbf.to_string())
```

```python
# ── Boxplot mit Streuung: Laufzeit-Ratio ─────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(16, 6), sharey=True)

for ax, solver in zip(axes, ["depqbf", "caqe"]):
    col = f"{solver}_time_ratio"
    ops = [op for op in OP_ORDER if op in df["operator"].cat.categories
           and len(df[df["operator"]==op][col].dropna()) > 0]

    data   = [df[df["operator"]==op][col].dropna().values for op in ops]
    colors = [COLORS[op] for op in ops]
    labels_short = [LABELS[op] for op in ops]

    bp = ax.boxplot(
        data, patch_artist=True, notch=False,
        medianprops={"color":"black","linewidth":2.5},
        whiskerprops={"linewidth":1.5,"linestyle":"--"},
        capprops={"linewidth":1.5},
        flierprops={"marker":"o","markersize":5,"alpha":0.6},
    )
    for patch, color in zip(bp["boxes"], colors):
        patch.set_facecolor(color); patch.set_alpha(0.75)
    for flier, color in zip(bp["fliers"], colors):
        flier.set_markerfacecolor(color)

    # Median als Text
    for i, d in enumerate(data):
        med = np.median(d)
        ax.text(i+1, med+0.005, f"{med:.3f}",
                ha="center", va="bottom", fontsize=8, fontweight="bold")

    ax.axhline(1.0, color="crimson", linestyle="--", linewidth=1.5,
               label="Ratio = 1.0 (kein Effekt)")
    ax.set_xticks(range(1, len(ops)+1))
    ax.set_xticklabels(labels_short, rotation=30, ha="right", fontsize=9)
    ax.set_ylabel("Laufzeit-Ratio (Mutant / Original)" if solver=="depqbf" else "")
    ax.set_title(f"Laufzeit-Ratio — {solver.upper()}", fontweight="bold")
    ax.legend(fontsize=9); ax.grid(axis="y", linestyle="--", alpha=0.4)

plt.suptitle("Primäranalyse: Laufzeit-Ratio pro Operator\n"
             "(Median = schwarze Linie, Punkte = Ausreißer)",
             fontsize=13, fontweight="bold")
plt.tight_layout()
plt.savefig("results/plots/boxplot_ratio.png", dpi=300, bbox_inches="tight")
plt.show()
print("✓ Gespeichert: results/plots/boxplot_ratio.png")
```

## 4. Primäranalyse — Coverage-Delta

**Kennzahl:** `delta_coverage = coverage_mutant - coverage_original`  
Zeigt welche Solver-Code-Pfade durch die Transformation zusätzlich aktiviert werden.

```python
# ── Deskriptive Statistik Coverage ──────────────────────────────────────
cov_stats = df.groupby("operator", observed=True).agg(
    n=("line_coverage_delta","count"),
    median_line=("line_coverage_delta","median"),
    std_line=("line_coverage_delta","std"),
    median_branch=("branch_coverage_delta","median"),
    std_branch=("branch_coverage_delta","std"),
).round(3)
cov_stats.index = cov_stats.index.map(LABELS)
print("Coverage-Delta Statistik:")
print(cov_stats.to_string())
```

```python
# ── Balkendiagramm: Delta-Coverage ──────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(16, 6), sharey=False)

for ax, ctype in zip(axes, ["line", "branch"]):
    col = f"{ctype}_coverage_delta"
    ops = [op for op in OP_ORDER if op in df["operator"].cat.categories]

    medians = [df[df["operator"]==op][col].median() for op in ops]
    stds    = [df[df["operator"]==op][col].std() for op in ops]
    colors  = [COLORS[op] for op in ops]
    labels_short = [LABELS[op] for op in ops]
    x = np.arange(len(ops))

    bars = ax.bar(x, medians, color=colors, alpha=0.82,
                  yerr=stds, capsize=5,
                  error_kw={"elinewidth":1.5,"ecolor":"black","alpha":0.6},
                  zorder=3)

    for bar, val in zip(bars, medians):
        ax.text(bar.get_x()+bar.get_width()/2, val+1,
                f"+{val:.1f}%", ha="center", va="bottom",
                fontsize=8, fontweight="bold")

    ax.axhline(0, color="crimson", linestyle="--", linewidth=1.5, label="Delta=0")
    ax.set_xticks(x)
    ax.set_xticklabels(labels_short, rotation=30, ha="right", fontsize=9)
    ax.set_ylabel("Delta-Coverage (Prozentpunkte)")
    ax.set_title(f"Delta {ctype.capitalize()}-Coverage", fontweight="bold")
    ax.legend(fontsize=9); ax.grid(axis="y", linestyle="--", alpha=0.4, zorder=0)

plt.suptitle("Primäranalyse: Coverage-Delta pro Operator\n"
             "(Median + Standardabweichung, nur äquivalente Transformationen)",
             fontsize=13, fontweight="bold")
plt.tight_layout()
plt.savefig("results/plots/coverage_delta.png", dpi=300, bbox_inches="tight")
plt.show()
print("✓ Gespeichert: results/plots/coverage_delta.png")
```

## 5. Sekundäranalyse — Mechanismen & Solver-Vergleich

**Frage:** Reagieren DepQBF und CAQE gleich auf dieselben Transformationen?  
→ Gleiche Reaktion = allgemeiner Effekt, nicht solver-spezifisch

```python
# ── Solver-Vergleich: DepQBF vs. CAQE ───────────────────────────────────
ops = [op for op in OP_ORDER if op in df["operator"].cat.categories]
labels_short = [LABELS[op] for op in ops]
x = np.arange(len(ops))
width = 0.38

fig, ax = plt.subplots(figsize=(14, 6))

d_medians = [df[df["operator"]==op]["depqbf_time_ratio"].median() for op in ops]
c_medians = [df[df["operator"]==op]["caqe_time_ratio"].median()   for op in ops]

bars1 = ax.bar(x-width/2, d_medians, width, label="DepQBF",
               color="#2d7dd2", alpha=0.85, zorder=3)
bars2 = ax.bar(x+width/2, c_medians, width, label="CAQE",
               color="#ef8a17", alpha=0.85, zorder=3)

for bars, vals in [(bars1, d_medians), (bars2, c_medians)]:
    for bar, val in zip(bars, vals):
        ax.text(bar.get_x()+bar.get_width()/2, val+0.001,
                f"{val:.3f}", ha="center", va="bottom", fontsize=8, fontweight="bold")

ax.axhline(1.0, color="crimson", linestyle="--", linewidth=1.5,
           label="Kein Unterschied (Ratio = 1.0)")
ax.set_xticks(x)
ax.set_xticklabels(labels_short, rotation=30, ha="right", fontsize=9)
ax.set_ylabel("Laufzeit-Ratio (Median)")
ax.set_title("Sekundäranalyse: DepQBF vs. CAQE — Reagieren beide Solver gleich?\n"
             "(Ähnliche Balken = allgemeiner Effekt, nicht solver-spezifisch)",
             fontweight="bold")
ax.legend(fontsize=10); ax.grid(axis="y", linestyle="--", alpha=0.4, zorder=0)
plt.tight_layout()
plt.savefig("results/plots/solver_comparison.png", dpi=300, bbox_inches="tight")
plt.show()
print("✓ Gespeichert: results/plots/solver_comparison.png")
```

## 6. Sekundäranalyse — Zusammenhang Coverage vs. Laufzeit

**Frage:** Erklärt mehr aktivierter Solver-Code (höhere Coverage) auch längere Laufzeit?  
→ Wenn ja: Trendlinie steigt → Coverage-Aktivierung kostet Zeit  
→ Wenn nein: Trendlinie flach → Coverage und Laufzeit sind unabhängig

```python
# ── Scatter: Coverage-Delta vs. Laufzeit-Ratio ──────────────────────────
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()

groups = {
    "Tautologie-Einfügung":    ["tautology_insertion"],
    "Pure Literal Elimination":["pure_literal_gen_auto","pure_literal_gen_pos","pure_literal_gen_neg"],
    "Universal Reduction":     ["universal_reduction_trigger"],
    "Blocked Clause Addition": ["blocked_clause_insertion","blocked_clause_insertion_neg","blocked_clause_insertion_v1"],
}
group_colors = {
    "Tautologie-Einfügung":    "#4C72B0",
    "Pure Literal Elimination":"#DD8452",
    "Universal Reduction":     "#55A868",
    "Blocked Clause Addition": "#C44E52",
}

for ax, (group_name, ops) in zip(axes, groups.items()):
    sub = df[df["operator"].isin(ops)]
    color = group_colors[group_name]

    if sub.empty:
        ax.text(0.5, 0.5, "Keine Daten", ha="center", va="center",
                transform=ax.transAxes, color="gray"); continue

    x_vals = sub["line_coverage_delta"]
    y_vals = sub["depqbf_time_ratio"]

    # Verschiedene Marker pro Operator-Variante
    for i, op in enumerate(ops):
        s = sub[sub["operator"]==op]
        markers = ["o","s","^"]
        ax.scatter(s["line_coverage_delta"], s["depqbf_time_ratio"],
                   c=color, alpha=0.75, s=55, zorder=3,
                   marker=markers[i % 3], label=LABELS.get(op, op),
                   edgecolors="white", linewidths=0.5)

    # Trendlinie + Korrelation
    if len(x_vals.dropna()) > 3:
        idx = x_vals.dropna().index.intersection(y_vals.dropna().index)
        slope, intercept, r, p, _ = stats.linregress(x_vals[idx], y_vals[idx])
        x_line = np.linspace(x_vals.min(), x_vals.max(), 50)
        ax.plot(x_line, slope*x_line+intercept,
                color=color, linewidth=2, linestyle="--", alpha=0.9)
        ax.text(0.05, 0.95, f"r={r:.3f}, p={p:.3f}\nm={slope:.5f}",
                transform=ax.transAxes, fontsize=9, va="top",
                bbox=dict(boxstyle="round", facecolor="white", alpha=0.8))

    ax.axhline(1.0, color="crimson", linestyle=":", linewidth=1.0, alpha=0.6)
    ax.set_xlabel("Δ Line-Coverage (%)")
    ax.set_ylabel("Laufzeit-Ratio DepQBF")
    ax.set_title(group_name, fontweight="bold", color=color)
    ax.legend(fontsize=8, loc="upper right"); ax.grid(linestyle="--", alpha=0.3)

plt.suptitle("Sekundäranalyse: Erklärt Coverage-Änderung die Laufzeit-Änderung?\n"
             "(r = Pearson-Korrelation, m = Steigung der Trendlinie)",
             fontsize=13, fontweight="bold")
plt.tight_layout()
plt.savefig("results/plots/scatter_coverage_vs_ratio.png", dpi=300, bbox_inches="tight")
plt.show()
print("✓ Gespeichert: results/plots/scatter_coverage_vs_ratio.png")
```

## 7. Aggregierte Ergebnistabelle (Thesis-fertig)

Vollständige Übersicht aller Operatoren mit statistischen Kennzahlen.

```python
# ── Vollständige Ergebnistabelle ─────────────────────────────────────────
rows = []
for op in OP_ORDER:
    sub = df[df["operator"] == op]
    if sub.empty:
        continue
    n   = len(sub)
    rd  = sub["depqbf_time_ratio"]
    rc  = sub["caqe_time_ratio"]
    dl  = sub["line_coverage_delta"]
    db  = sub["branch_coverage_delta"]

    # Effekt-Interpretation
    med_rd = rd.median()
    if med_rd < 0.995:
        effekt = "↓ schneller"
    elif med_rd > 1.005:
        effekt = "↑ langsamer"
    else:
        effekt = "→ neutral"

    # Wilcoxon-Test: Ist Ratio signifikant != 1.0?
    try:
        _, p_val = stats.wilcoxon(rd - 1.0)
        sig = "**" if p_val < 0.01 else ("*" if p_val < 0.05 else "n.s.")
    except Exception:
        sig = "—"

    rows.append({
        "Operator":           LABELS.get(op, op),
        "n":                  n,
        "Ratio DepQBF (Median)":  f"{rd.median():.4f} ±{rd.std():.4f}",
        "Ratio CAQE (Median)":    f"{rc.median():.4f} ±{rc.std():.4f}",
        "Δ Line-Cov. (Median)":   f"+{dl.median():.2f}%",
        "Δ Branch-Cov. (Median)": f"+{db.median():.2f}%",
        "Aktivierter Mechanismus": MECHANISMS.get(op, "—"),
        "Effekt":             effekt,
        "Signifikanz (Wilcoxon)": sig,
    })

result_df = pd.DataFrame(rows).set_index("Operator")
pd.set_option("display.max_colwidth", 35)
pd.set_option("display.width", 200)
print(result_df.to_string())
```

```python
# ── Schoen formatierte Tabelle mit Styling ──────────────────────────────
def color_effekt(val):
    if "schneller" in str(val): return "color: green; font-weight: bold"
    if "langsamer" in str(val): return "color: red; font-weight: bold"
    return ""

def color_ratio(val):
    try:
        v = float(str(val).split()[0])
        if v < 0.995: return "background-color: #d4efdf"
        if v > 1.005: return "background-color: #fadbd8"
    except:
        pass
    return ""

styled = result_df.style \
    .applymap(color_effekt, subset=["Effekt"]) \
    .applymap(color_ratio, subset=["Ratio DepQBF\n(Median)","Ratio CAQE\n(Median)"]) \
    .set_caption("Aggregierte Ergebnistabelle — QBF Transformer Experimente") \
    .set_properties(**{"text-align": "center", "font-size": "11px"}) \
    .set_table_styles([
        {"selector":"th", "props":[("background-color","#2d3a4a"),("color","white"),
                                    ("font-weight","bold"),("text-align","center")]},
        {"selector":"caption","props":[("font-size","14px"),("font-weight","bold"),
                                        ("margin-bottom","10px")]},
    ])
styled
```

## 8. Statistische Tests

Ist der Laufzeit-Effekt statistisch signifikant oder nur Rauschen?

```python
# ── Wilcoxon-Vorzeichen-Rang-Test ─────────────────────────────────────────
# H0: Median-Ratio = 1.0 (kein Effekt)
# H1: Median-Ratio != 1.0 (signifikanter Effekt)
print("Wilcoxon-Test: Ist Laufzeit-Ratio signifikant != 1.0?")
print(f"{'Operator':<30} {'n':>4}  {'Median':>8}  {'p-Wert':>10}  {'Signifikanz'}")
print("-" * 70)

for op in OP_ORDER:
    sub = df[df["operator"]==op]["depqbf_time_ratio"].dropna()
    if len(sub) < 3:
        print(f"  {LABELS.get(op,op):<28} {len(sub):>4}  {'—':>8}  {'—':>10}  zu wenig Daten")
        continue
    try:
        stat, p = stats.wilcoxon(sub - 1.0)
        sig = "*** p<0.001" if p<0.001 else ("** p<0.01" if p<0.01 else ("* p<0.05" if p<0.05 else "n.s."))
        print(f"  {LABELS.get(op,op):<28} {len(sub):>4}  {sub.median():>8.4f}  {p:>10.4f}  {sig}")
    except Exception as e:
        print(f"  {LABELS.get(op,op):<28} {len(sub):>4}  {sub.median():>8.4f}  {'Fehler':>10}")

print()
print("Legende: *** = p<0.001,  ** = p<0.01,  * = p<0.05,  n.s. = nicht signifikant")
```

```python
# ── Kruskal-Wallis-Test: Unterscheiden sich Operatoren untereinander? ──
# H0: Alle Operatoren haben dieselbe Laufzeit-Ratio-Verteilung

groups_data = [
    df[df["operator"]==op]["depqbf_time_ratio"].dropna().values
    for op in OP_ORDER
    if op in df["operator"].values and len(df[df["operator"]==op]) >= 3
]
valid_ops = [
    op for op in OP_ORDER
    if op in df["operator"].values and len(df[df["operator"]==op]) >= 3
]

if len(groups_data) >= 2:
    stat, p = stats.kruskal(*groups_data)
    print(f"Kruskal-Wallis-Test (alle Operatoren):")
    print(f"  H = {stat:.4f},  p = {p:.6f}")
    if p < 0.05:
        print(f"  → SIGNIFIKANT: Die Operatoren unterscheiden sich in ihrer")
        print(f"    Wirkung auf die Laufzeit (p < 0.05)")
    else:
        print(f"  → nicht signifikant (p >= 0.05)")
else:
    print("Nicht genug Gruppen fuer Kruskal-Wallis-Test")
```

## 9. Zusammenfassung & alle Plots speichern

```python
from pathlib import Path
Path("results/plots").mkdir(parents=True, exist_ok=True)

print("=" * 60)
print("ZUSAMMENFASSUNG")
print("=" * 60)
print(f"Datensatz: {len(df_raw)} Experimente gesamt")
print(f"           {len(df)} gueltig & aequivalent ({len(df)/len(df_raw)*100:.1f}%)")
print()
print("Top-Ergebnisse:")

best_speedup = df.groupby("operator", observed=True)["depqbf_time_ratio"].median().idxmin()
best_coverage = df.groupby("operator", observed=True)["line_coverage_delta"].median().idxmax()

print(f"  Groesste Beschleunigung: {LABELS.get(best_speedup,best_speedup)}")
print(f"    Median Ratio = {df[df['operator']==best_speedup]['depqbf_time_ratio'].median():.4f}")
print()
print(f"  Groesste Coverage-Aktivierung: {LABELS.get(best_coverage,best_coverage)}")
print(f"    Median Delta = +{df[df['operator']==best_coverage]['line_coverage_delta'].median():.2f}%")
print()
print("Gespeicherte Plots:")
for p in sorted(Path("results/plots").glob("*.png")):
    print(f"  {p}")
```
